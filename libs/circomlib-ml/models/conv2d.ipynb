```python
from tensorflow.keras.layers import Input, Conv2D
from tensorflow.keras import Model
import numpy as np
```

```python
inputs = Input(shape=(5,5,3))
x = Conv2D(2, 3)(inputs)
model = Model(inputs, x)
```

```python
model.summary()
```

```python
model.weights
```

```python
X = np.random.rand(1,5,5,3)
X
```

```python
y = model.predict(X)
y
```

```python
in_json = {
    "in": (X[0]*1000).round().astype(int).astype(str).tolist(),
    "weights": (model.weights[0].numpy()*1000).round().astype(int).astype(str).tolist(),
    "bias": ["0","0"]
}
```

```python
out_json = {
    "out": (y*1000000).round().astype(int).flatten().tolist()
}
```

```python
import json
```

```python
with open("conv2D_input.json", "w") as f:
    json.dump(in_json, f)
```

```python
with open("conv2D_output.json", "w") as f:
    json.dump(out_json, f)
```
