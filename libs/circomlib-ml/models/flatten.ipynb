```python
from tensorflow.keras.layers import Input, Flatten
from tensorflow.keras import Model
import numpy as np
```

```python
inputs = Input(shape=(5,5,3))
x = Flatten()(inputs)
model = Model(inputs, x)
```

```python
model.summary()
```

```python
model.weights
```

```python
X = np.random.rand(5,5,3)
X
```

```python
# Flatten doesn't do any prediction
# y = model.predict(X)
y = X.flatten()
y
```

```python
in_json = {
    "in": (X*1000).round().astype(int).astype(str).tolist()
}
```

```python
out_json = {
    "out": (y*1000).round().astype(int).astype(str).flatten().tolist()
}
```

```python
import json
```

```python
with open("flatten2D_input.json", "w") as f:
    json.dump(in_json, f, indent=4)
```

```python
with open("flatten2D_output.json", "w") as f:
    json.dump(out_json, f, indent=4)
```
