```python
from tensorflow.keras.layers import Input, Conv1D
from tensorflow.keras import Model
import numpy as np
```

```python
inputs = Input(shape=(20,3))
x = Conv1D(2,4,3)(inputs)
model = Model(inputs, x)
```

```python
model.summary()
```

```python
model.weights
```

```python
X = np.random.rand(1,20,3)
X
```

```python
y = model.predict(X)
y
```

```python
# Unwrap X and y out of their containing single-element arrays for circom consumption
X = X[0]
y = y[0]
```

```python
in_json = {
    "in": (X*1000).round().astype(int).astype(str).tolist(),
    "weights": (model.weights[0].numpy()*1000).round().astype(int).astype(str).tolist(),
    "bias": np.zeros(model.weights[1].numpy().shape).astype(int).astype(str).tolist()
}
in_json
```

```python
out_json = {
    "out": (y*1000000).round().astype(int).astype(str).tolist()
}
out_json
```

```python
in_flat_json = {
    "in": (X*1000).round().astype(int).flatten().tolist(),
    "weights": (model.weights[0].numpy()*1000).round().astype(int).flatten().tolist(),
    "bias": np.zeros(model.weights[1].numpy().shape).tolist()
}
in_flat_json
```

```python
out_flat_json = {
    "out": (y*1000000).round().astype(int).flatten().tolist()
}
out_flat_json
```

```python
import json
```

```python
with open("conv1D_input.json", "w") as f:
    json.dump(in_json, f, indent=4)
```

```python
with open("conv1D_output.json", "w") as f:
    json.dump(out_json, f, indent=4)
```

```python
with open("conv1D_input_flatten.json", "w") as f:
    json.dump(in_flat_json, f, indent=4)
```

```python
with open("conv1D_output_flattened.json", "w") as f:
    json.dump(out_flat_json, f, indent=4)
```
